```python
from typing import TypedDict, List
from langgraph.graph import StateGraph, START, END

# 1. Define the State
# This holds the data as it passes through the Hive-Mind
class AgentState(TypedDict):
    problem_statement: str
    sub_hypotheses: List[str]
    retrieved_papers: List[dict]
    evaluated_papers: List[dict]

# 2. Define the Agents (Nodes)

def orchestrator_node(state: AgentState):
    """
    The Orchestrator: Acts as the Principal Investigator. 
    Deconstructs the PS into specific research hypotheses.
    """
    print("--- ORCHESTRATOR: Deconstructing Problem Statement ---")
    ps = state["problem_statement"]
    
    # Mocking the LLM splitting the PS into sub-queries
    mock_hypotheses = [
        f"State-of-the-art architectures for {ps[:10]}...",
        f"Feasibility of mathematical preservation in {ps[:10]}..."
    ]
    return {"sub_hypotheses": mock_hypotheses}

def librarian_node(state: AgentState):
    """
    The Librarian: Interacts with academic APIs to fetch papers.
    """
    print("--- LIBRARIAN: Fetching research papers ---")
    queries = state["sub_hypotheses"]
    
    # Mocking the recursive link-following from ArXiv/Semantic Scholar
    mock_papers = [
        {"title": "GraphRAG Basics", "content": "Graph-based RAG overview", "source": "ArXiv"},
        {"title": "Nougat PDF parsing", "content": "Mathematical preservation in PDFs", "source": "Semantic Scholar"}
    ]
    return {"retrieved_papers": mock_papers}

def critic_node(state: AgentState):
    """
    The Critic: Evaluates paper quality based on relevance to the PS.
    """
    print("--- CRITIC: Evaluating relevance ---")
    papers = state["retrieved_papers"]
    
    # Mocking the evaluation metric (mathematical relevance)
    evaluated = []
    for paper in papers:
        paper["relevance_score"] = 0.95 if "Graph" in paper["title"] else 0.75
        if paper["relevance_score"] > 0.80:
            evaluated.append(paper)
            
    return {"evaluated_papers": evaluated}

# 3. Build the LangGraph Routing

workflow = StateGraph(AgentState)

# Add the nodes
workflow.add_node("orchestrator", orchestrator_node)
workflow.add_node("librarian", librarian_node)
workflow.add_node("critic", critic_node)

# Define the edges (The Workflow)
workflow.add_edge(START, "orchestrator")
workflow.add_edge("orchestrator", "librarian")
workflow.add_edge("librarian", "critic")
workflow.add_edge("critic", END)

# Compile the graph
app = workflow.compile()

# --- 4. Run the Prototype ---
if __name__ == "__main__":
    initial_state = {
        "problem_statement": "Develop an automated research synthesis tool for technical competitions.",
        "sub_hypotheses": [],
        "retrieved_papers": [],
        "evaluated_papers": []
    }
    
    print("Starting Agentic Framework Pipeline...\n")
    result = app.invoke(initial_state)
    
    print("\n--- FINAL OUTPUT ---")
    print(f"Top Relevant Papers for IITG.AI PS: {len(result['evaluated_papers'])}")
    for p in result['evaluated_papers']:
        print(f"- {p['title']} (Score: {p['relevance_score']})")
```
