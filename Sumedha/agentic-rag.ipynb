```python
import chromadb
import os
from sentence_transformers import SentenceTransformer
from groq import Groq
from kaggle_secrets import UserSecretsClient
secret = UserSecretsClient()
api_key = secret.get_secret("GROQ_API_KEY")

groq_client = Groq(api_key=api_key)

# ── Setup ──────────────────────────────────────────────
os.makedirs("/kaggle/working/chroma_db", exist_ok=True)
client = chromadb.PersistentClient(path="/kaggle/working/chroma_db")
collection = client.get_or_create_collection(name="papers")
model = SentenceTransformer("allenai-specter")



from marker.converters.pdf import PdfConverter
from marker.models import create_model_dict
from marker.config.parser import ConfigParser

config = ConfigParser({})
models = create_model_dict()
converter = PdfConverter(config=config.generate_config_dict(), artifact_dict=models)

def pdf_to_markdown(pdf_path):
    rendered = converter(pdf_path)
    return rendered.markdown
# ── Step 2: Chunk ──────────────────────────────────────
import re

def chunk_text(text, chunk_size=10, min_tokens=30):
    sentences = re.split(r'(?<=[.!?]) +', text)
    chunks = []
    for i in range(0, len(sentences), chunk_size):
        chunk = " ".join(sentences[i:i+chunk_size])
        if len(chunk.split()) >= min_tokens:
            chunks.append(chunk)
    return chunks

# ── Step 3: Embed + Store in ChromaDB ─────────────────
def index_paper(pdf_path, paper_id):
    text = pdf_to_markdown(pdf_path)
    chunks = chunk_text(text)
    embeddings = model.encode(chunks).tolist()
    
    collection.add(
        ids=[f"{paper_id}_chunk_{i}" for i in range(len(chunks))],
        embeddings=embeddings,
        documents=chunks,
        metadatas=[{"paper_id": paper_id} for _ in chunks]
    )
    print(f"Indexed {len(chunks)} chunks from {paper_id}")

# ── Step 4: Retrieve ───────────────────────────────────
def retrieve(query, top_k=5):
    query_vec = model.encode(query).tolist()
    results = collection.query(
        query_embeddings=[query_vec],
        n_results=top_k
    )
    return results["documents"][0]

# ── Step 5: Generate Answer ────────────────────────────
def answer(query):
    chunks = retrieve(query)
    context = "\n\n".join(chunks)
    
    response = groq_client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[{
            "role": "user",
            "content": f"""Answer based only on the context provided.

Context:
{context}

Question: {query}"""
        }]
    )
    return response.choices[0].message.content

# ── Usage ──────────────────────────────────────────────
index_paper("/kaggle/input/datasets/sumedha0711/new123/attention-is-all-you-need.pdf", "paper_001")
print(answer("What attention mechanism does the Transformer use?"))
```
