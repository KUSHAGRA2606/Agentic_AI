**Chunking**

Sentence based chunking with overlap  
later try token based chunking

```python
def chunk_text(text, sentences_per_chunk=5, overlap=2):  #overlap is basically overlap btw chunks

    # Splitting text into sentences
    sentences = re.split(r'(?<=[.!?])\s+', text)

    chunks = []
    step = sentences_per_chunk - overlap
    for i in range(0, len(sentences), step):
        chunk = " ".join(sentences[i:i + sentences_per_chunk])
        # Avoid empty chunks
        if chunk.strip():
            chunks.append(chunk)

    return chunks
```

```python
chunks=chunk_text(markdown, 5, 1)
```

**VectorDB creation and storage and indexing**

```python
client = chromadb.PersistentClient(path="./chroma_db")  #clien basically acts as our interface to the db
#persistent client is basically so that data is stored permanently instead of in temporary ram(this is second option which cant be used here)
#path="./chroma_db" stores db files in this folder

collection_specter = client.get_or_create_collection(name="papers_specter") #collection is basically a container/table for the embeddings
collection_bge = client.get_or_create_collection(name="papers_bge")
#always 1 collection for 1 embedding model dont just change the embedding model in same var and run again

ids = [f"chunk_{i}"
       for i in range(len(chunks))
]

metadatas = [
    {
        "paper": "RAG_research_paper",  #later for multiple papers we will use a nested loop to change the paper number also and we will need a list of papers along with the list of chunks for each paper
        "chunk_id": i
    }
    for i in range(len(chunks))
]

collection_bge.add(ids=ids,    #actual indexing part, we store everything we need later to retrieve or give the llm or search or anything else
               embeddings=chunk_embeddings.tolist(), 
               documents=chunks, 
               metadatas=metadatas
              )

print(collection_bge.count())

```

**Retrieval**

here we will embed the query then use similarity search to get most matching chunks

instead of below also try to get like the top say 10 above a threshold for the papers so if there are none or only 1-2 maybe the paper is not good or it might work as a better retrieval method  
if u do this also see how to optimize th threshold and change n_results in this func appropriately

```python
def retrieve(query, top_k=5):

    # Convert query into embedding
    query_embedding = model.encode(query).tolist()

    # Search vector DB
    results = collection_bge.query(query_embeddings=[query_embedding], n_results=top_k)
    #collection.query is the actual search function using cosine similarity
    
    return results["documents"][0]  #[0] returns results for query 1, for multiple queries we will feed it [1], [2] ...
```

**Answer Generation**

```python
def answer(query, top_k=10):

    #retrieval
    chunks = retrieve(query, top_k)

    context = "\n".join(chunks)

    response = client.chat.completions.create(
        model="llama-3.3-70b-versatile",
        messages=[
            {
                "role": "system",    #this sets the rules for the llm, it will first see this before question and context
                "content": "Answer the question ONLY using the provided context. Do not use any outside knowledge. If the answer is not explicitly present in the context, respond with: 'Answer not found in provided context.'"
            },
            {
                "role": "user",
                "content": f"""
Context:
{context}

Question:
{query}
"""
            }
        ],
        temperature=0.2
    )

    return response.choices[0].message.content
```
